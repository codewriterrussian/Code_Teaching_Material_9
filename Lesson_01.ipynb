# Lesson 01｜Python 基礎輸入與輸出

本課介紹程式的基本運作方式，以及 Python 的 `input()` 和 `print()`。

## 學習目標

- 理解程式的輸入、處理與輸出。
- 使用 `print()` 顯示文字與多個值。
- 使用 `input()` 接收使用者輸入。
- 使用 `sep` 控制多個輸出項目之間的分隔符號。

> 請依序執行下方的程式碼儲存格。遇到 `input()` 時，輸入文字後按 Enter 即可繼續執行。

## 1. 程式的三個基本部分

| 部分 | 英文 | 說明 |
|---|---|---|
| 輸入 | Input | 接收程式需要的資料。 |
| 處理 | Processing | 依照程式指令處理資料或進行計算。 |
| 輸出 | Output | 顯示或傳回處理結果。 |

例如，計算程式接收數字、進行計算，再輸出結果；日誌程式也可能不需要使用者手動輸入，而是直接處理資料並輸出紀錄。

## 2. 使用 `print()` 輸出文字

`print()` 可以將指定的內容顯示在畫面上。文字必須使用單引號或雙引號括起來。

**預期輸出：**

```text
hi
```

````python
print("hi")
````

## 3. 使用 `input()` 接收輸入

`input()` 會等待使用者輸入資料，按下 Enter 後傳回輸入的文字（字串）。

以下範例沒有儲存或使用輸入的內容。單獨呼叫 `print()` 不會顯示任何文字，只會輸出一個換行。

**操作方式：** 輸入任意文字後按 Enter，觀察程式的結果。

````python
input()
print()
````

## 4. 將輸入內容直接輸出

可以將 `input()` 放入 `print()` 的括號中，讓程式先接收使用者輸入，再將輸入的內容顯示出來。

**範例輸入：**

```text
hi
```

**預期輸出：**

```text
hi
```

````python
print(input())
````

## 5. 同時輸出多個項目

在 `print()` 中，可以使用逗號分隔多個項目。預設情況下，Python 會在每兩個輸出項目之間加入一個空格。

**預期輸出：**

```text
a b
```

````python
print("a", "b")
````

### 延伸範例：結合固定文字與使用者輸入

輸入 `Amy` 時，程式會輸出 `hello, Amy`。注意逗號是第一個字串的一部分，而兩個輸出項目之間的空格由 `print()` 自動加入。

````python
print("hello,", input())
````

## 6. 使用 `sep` 自訂分隔符號

`sep` 是 `print()` 的選用參數，用來指定多個輸出項目之間的分隔符號。

| 寫法 | 輸出結果 |
|---|---|
| `print("a", "b")` | `a b` |
| `print("a", "b", sep=",")` | `a,b` |
| `print("a", "b", sep="")` | `ab` |

以下範例將預設的空格改為逗號。

````python
print("a", "b", sep=",")
````

### 綜合範例：輸出加油訊息

使用 `sep=""` 讓多個輸出項目直接接在一起，並在固定字串中保留需要的標點符號及空格。

**範例輸入：**

```text
Amy
```

**預期輸出：**

```text
Go, Amy, go go
```

````python
print("Go, ", input(), ", go go", sep="")
````

## 7. 課堂重點整理

| 語法 | 用途 |
|---|---|
| `input()` | 接收一行使用者輸入，傳回字串。 |
| `print()` | 輸出一個換行。 |
| `print("hi")` | 輸出指定文字。 |
| `print(input())` | 直接輸出使用者輸入的文字。 |
| `print("a", "b")` | 同時輸出多個項目，預設以空格分隔。 |
| `print("a", "b", sep=",")` | 使用逗號分隔輸出項目。 |
| `print("a", "b", sep="")` | 不加入額外的分隔符號。 |
